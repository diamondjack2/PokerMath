```python
def compare_hands(hands:list) -> list:
    '''Takes a list of hands that have been evaluated and returns a list of the winning hands index'''
    winning_pos = []
    winning_hand = max(hands)
    for idx, hand in enumerate(hands):
        if hand == winning_hand:
            winning_pos.append(idx)
    return winning_pos
```

```python
test_1 = [(2,9,11,7,5), (2,9,11,7,3)]
compare_hands(test_1)
```

```python
from treys import Card, Evaluator

treys_eval = Evaluator()

def treys_verdict(board, hole_1, hole_2):
    board_t = [Card.new(c) for c in board]
    s1 = treys_eval.evaluate(board_t, [Card.new(c) for c in hole_1])
    s2 = treys_eval.evaluate(board_t, [Card.new(c) for c in hole_2])
    if s1 < s2:
        return [0]
    elif s2 < s1:
        return [1]
    else:
        return [0, 1]
```

```python
treys_list = []
my_list = []


for i in range(100000):
    shuffle = random.sample(cards.DECK, 9)
    board = shuffle[:5]
    
    p1_hole = shuffle[5:7]
    p2_hole = shuffle[7:]
    
    p1_hand = board + p1_hole
    p2_hand = board + p2_hole
    
    p1_str = evaluator.evaluate_hand(p1_hand)
    p2_str = evaluator.evaluate_hand(p2_hand)

    table = [p1_str, p2_str]
    
    verdict = compare_hands(table)

    my_list.append(verdict)

    treys_v = treys_verdict(board,p1_hole,p2_hole)

    treys_list.append(treys_v)


treys_list == my_list
```
